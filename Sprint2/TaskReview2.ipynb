```
using System;

public interface IReportable 
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string [] GetTeamMembers();
}

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public decimal Salary { get; set; }

    public Employee(string name, string position, decimal salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Должность: {Position}, Зарплата: {Salary} руб.");
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, decimal salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    private string[] teamMembers;

    public ProjectManager(string name, string position, decimal salary,
                          string department, string projectName, string[] teamMembers)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        this.teamMembers = teamMembers;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"Отчёт по проекту «{ProjectName}»: руководитель {Name}, отдел {Department}, команда из {teamMembers.Length} чел.";
    }

    public string[] GetTeamMembers()
    {
        return teamMembers;
    }
}


Console.WriteLine("1. Employee ");
Employee emp = new Employee("Мария Паутова", "Разработчик", 80000);
emp.PrintInfo();

Console.WriteLine("\n2. Manager ");
Manager mgr = new Manager("Гуринчук Анастасия", "Менеджер", 120000, "IT-отдел");
mgr.PrintInfo();

Console.WriteLine("\n3. ProjectManager ");
ProjectManager pm = new ProjectManager(
    "Молоканов Никита", "Проектный менеджер", 180000,
    "Отдел разработки", "Внедрение CRM",
    new string[] { "Мария Паутова", "Гуринчук Анастасия", "Молоканов Никита" }
);
pm.PrintInfo();

Console.WriteLine("\n4. IReportable ");
Console.WriteLine(pm.GenerateReport());

Console.WriteLine("\n5. ITeamLeader ");
foreach (var member in pm.GetTeamMembers())
{
    Console.WriteLine($"  • {member}");
}
```
