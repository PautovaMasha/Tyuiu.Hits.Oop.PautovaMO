классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
using System;

public interface ITrackable
{
    string TrackNumber { get; set; }
    void Track();
}

public interface IInsurable
{
    decimal InsuranceCost { get; set; }
    void Insure();
}

public class ShippingOption
{
    public int Id { get; set; }
    public string Name { get; set; }
    public decimal Cost { get; set; }
    public string Carrier { get; set; }
    public double WeightKg { get; set; }

    public ShippingOption(int id, string name, decimal cost, string carrier, double weightKg)
    {
        Id = id;
        Name = name;
        Cost = cost;
        Carrier = carrier;
        WeightKg = weightKg;
    }

    public virtual decimal CalculateCost() => Cost;
    public virtual string EstimateDeliveryTime() => "Время не указано";
    public virtual string GetDetails() => $"ID: {Id}, {Name}, {Cost} руб., Перевозчик: {Carrier}, Вес: {WeightKg} кг";
    public virtual void PrintLabel() => Console.WriteLine($"Этикетка: [{Id}] {Name} ({Carrier})");
}

public class StandardDelivery : ShippingOption
{
    public int AvgDays { get; set; }
    public bool IsTrackable { get; set; }

    public StandardDelivery(int id, string name, decimal cost, string carrier, double weightKg,
                            int days, bool isTrackable)
        : base(id, name, cost, carrier, weightKg)
    {
        AvgDays = days;
        IsTrackable = isTrackable;
    }

    public override string EstimateDeliveryTime() => $"Среднее время: {AvgDays} дней";
    public override string GetDetails() => base.GetDetails() + $", {AvgDays} дней, отслеживание: {IsTrackable}";

    public void NotifyClient() => Console.WriteLine($"[{Name}] Клиент уведомлён о стандартной доставке");
}

public class ExpressDelivery : ShippingOption
{
    public int MinHours { get; set; }
    public decimal Surcharge { get; set; }
    public bool HasPriority { get; set; }

    public ExpressDelivery(int id, string name, decimal cost, string carrier, double weightKg,
                           int hours, decimal surcharge, bool hasPriority)
        : base(id, name, cost, carrier, weightKg)
    {
        MinHours = hours;
        Surcharge = surcharge;
        HasPriority = hasPriority;
    }

    public override decimal CalculateCost() => Cost + Surcharge;
    public override string EstimateDeliveryTime() => $"Минимальное время: {MinHours} часов";
    public override string GetDetails() =>
        base.GetDetails() + $", наценка {Surcharge} руб., итого {CalculateCost()} руб., приоритет: {HasPriority}";

    public void SendSms() => Console.WriteLine($"[{Name}] SMS отправлено клиенту об экспресс-доставке");
}

public class Pickup : ShippingOption
{
    public string Address { get; set; }
    public string WorkingHours { get; set; }

    public Pickup(int id, string name, decimal cost, string carrier, double weightKg,
                  string address, string workingHours)
        : base(id, name, cost, carrier, weightKg)
    {
        Address = address;
        WorkingHours = workingHours;
    }

    public override decimal CalculateCost() => 0;
    public override string EstimateDeliveryTime() => "Самовывоз — доступно сразу";
    public override string GetDetails() =>
        $"ID: {Id}, {Name}, Адрес: {Address}, Часы работы: {WorkingHours}";

    public void GiveDirections() => Console.WriteLine($"[{Name}] Клиент получил схему проезда к {Address}");
}

// ====== ЗАМЕНЕНО: было DroneDelivery, стало RobotCourierDelivery ======
public class RobotCourierDelivery : ExpressDelivery
{
    public double MaxRangeKm { get; set; }       // максимальная дальность робота
    public bool IsAutonomous { get; set; }       // автономный режим

    public RobotCourierDelivery(int id, string name, decimal cost, string carrier, double weightKg,
                                int hours, decimal surcharge, bool hasPriority,
                                double maxRangeKm, bool isAutonomous)
        : base(id, name, cost, carrier, weightKg, hours, surcharge, hasPriority)
    {
        MaxRangeKm = maxRangeKm;
        IsAutonomous = isAutonomous;
    }

    public override string GetDetails() =>
        base.GetDetails() + $", дальность: {MaxRangeKm} км, автономный: {IsAutonomous}";

    // ЗАМЕНЕНО: LaunchDrone → ActivateRobot
    public void ActivateRobot() => Console.WriteLine($"[{Name}] Робот-курьер активирован, дальность до {MaxRangeKm} км");
}

public class TrackedExpress : ExpressDelivery, ITrackable, IInsurable
{
    public string TrackNumber { get; set; }
    public decimal InsuranceCost { get; set; }

    public TrackedExpress(int id, string name, decimal cost, string carrier, double weightKg,
                          int hours, decimal surcharge, bool hasPriority,
                          string trackNumber, decimal insuranceCost)
        : base(id, name, cost, carrier, weightKg, hours, surcharge, hasPriority)
    {
        TrackNumber = trackNumber;
        InsuranceCost = insuranceCost;
    }

    public void Track() => Console.WriteLine($"[{Name}] Отслеживание по номеру: {TrackNumber}");

    public void Insure() => Console.WriteLine($"[{Name}] Застраховано на {InsuranceCost} руб.");

    public override decimal CalculateCost() => base.CalculateCost() + InsuranceCost;

    public override string GetDetails() =>
        base.GetDetails() + $", трек: {TrackNumber}, страховка: {InsuranceCost} руб.";
}

var deliveries = new ShippingOption[]
{
    new StandardDelivery(1, "Стандартная", 300, "Почта России", 2.5, 5, true),
    new ExpressDelivery(2, "Экспресс", 500, "СДЭК", 1.0, 6, 300, true),
    new Pickup(3, "Самовывоз", 0, "Своими силами", 0, "Москва, ул. Ленина, 10", "09:00–21:00"),
    new RobotCourierDelivery(4, "Робот-курьер", 800, "RoboExpress", 0.5, 2, 400, true, 15.0, true),
    new TrackedExpress(5, "Отслеживаемый экспресс", 600, "DHL", 1.5, 4, 250, true, "TRK-987654", 150)
};

foreach (var d in deliveries)
{
    Console.WriteLine(d.GetDetails());
    Console.WriteLine($"Время: {d.EstimateDeliveryTime()}");
    Console.WriteLine($"Стоимость: {d.CalculateCost()} руб.");
    d.PrintLabel();
    Console.WriteLine();
}

Console.WriteLine("=== Множественное наследование через интерфейсы ===");
TrackedExpress tracked = (TrackedExpress)deliveries[4];
tracked.Track();
tracked.Insure();
Console.WriteLine($"Итоговая стоимость со страховкой: {tracked.CalculateCost()} руб.");

public static class DeliveryInteractions
{
    public static void Compare(ShippingOption a, ShippingOption b)
    {
        Console.WriteLine($"Сравнение: {a.Name} vs {b.Name}");
        Console.WriteLine($"  Стоимость: {a.CalculateCost()} руб. vs {b.CalculateCost()} руб.");
        Console.WriteLine($"  Время: {a.EstimateDeliveryTime()} vs {b.EstimateDeliveryTime()}");

        if (a.CalculateCost() < b.CalculateCost())
            Console.WriteLine($"  Дешевле: {a.Name}");
        else if (a.CalculateCost() > b.CalculateCost())
            Console.WriteLine($"  Дешевле: {b.Name}");
        else
            Console.WriteLine("  Стоимость одинаковая");

        Console.WriteLine();
    }

    public static void RedirectToPickup(ShippingOption source, Pickup pickup)
    {
        Console.WriteLine($"{source.Name} перенаправляет клиента на самовывоз: {pickup.Address}");
        Console.WriteLine($"Итоговая стоимость: {pickup.CalculateCost()} руб.");
        Console.WriteLine($"Время: {pickup.EstimateDeliveryTime()}");
        Console.WriteLine();
    }
}

DeliveryInteractions.Compare(deliveries[0], deliveries[1]);
DeliveryInteractions.Compare(deliveries[1], deliveries[2]);
DeliveryInteractions.RedirectToPickup(deliveries[1], (Pickup)deliveries[2]);
```
