```
using System;

public class ShippingOption
{
    public int Id { get; set; }
    public string Name { get; set; }
    public decimal Cost { get; set; }

    public ShippingOption(int id, string name, decimal cost)
    {
        Id = id;
        Name = name;
        Cost = cost;
    }

    public virtual decimal CalculateCost() => Cost;
    public virtual string EstimateDeliveryTime() => "Время не указано";
    public virtual string GetDetails() => $"ID: {Id}, {Name}, {Cost} руб.";
}

public class StandardDelivery : ShippingOption
{
    public int AvgDays { get; set; }

    public StandardDelivery(int id, string name, decimal cost, int days) : base(id, name, cost)
    {
        AvgDays = days;
    }

    public override string EstimateDeliveryTime() => $"Среднее время: {AvgDays} дней";
    public override string GetDetails() => base.GetDetails() + $", {AvgDays} дней";
}

public class ExpressDelivery : ShippingOption
{
    public int MinHours { get; set; }
    public decimal Surcharge { get; set; }

    public ExpressDelivery(int id, string name, decimal cost, int hours, decimal surcharge) : base(id, name, cost)
    {
        MinHours = hours;
        Surcharge = surcharge;
    }

    public override decimal CalculateCost() => Cost + Surcharge;
    public override string EstimateDeliveryTime() => $"Минимальное время: {MinHours} часов";
    public override string GetDetails() => base.GetDetails() + $", наценка {Surcharge} руб., итого {CalculateCost()} руб.";
}

public class Pickup : ShippingOption
{
    public string Address { get; set; }

    public Pickup(int id, string name, decimal cost, string address) : base(id, name, cost)
    {
        Address = address;
    }

    public override decimal CalculateCost() => 0;
    public override string EstimateDeliveryTime() => "Самовывоз - доступно сразу";
    public override string GetDetails() => $"ID: {Id}, {Name}, {Cost} руб., Адрес: {Address}";
}

var deliveries = new ShippingOption[]
{
    new StandardDelivery(1, "Стандартная", 300, 5),
    new ExpressDelivery(2, "Экспресс", 500, 6, 300),
    new Pickup(3, "Самовывоз", 0, "Москва, ул. Ленина, 10")
};

foreach (var d in deliveries)
{
    display(d.GetDetails());
    display($"Время: {d.EstimateDeliveryTime()}");
    display($"Стоимость: {d.CalculateCost()} руб.");
    display("");
}

//  ДОБАВЛЕНО 
public static class DeliveryInteractions
{
    public static void Compare(ShippingOption a, ShippingOption b)
    {
        Console.WriteLine($"Сравнение: {a.Name} vs {b.Name}");
        Console.WriteLine($"  Стоимость: {a.CalculateCost()} руб. vs {b.CalculateCost()} руб.");
        Console.WriteLine($"  Время: {a.EstimateDeliveryTime()} vs {b.EstimateDeliveryTime()}");

        if (a.CalculateCost() < b.CalculateCost())
            Console.WriteLine($"  Дешевле: {a.Name}");
        else if (a.CalculateCost() > b.CalculateCost())
            Console.WriteLine($"  Дешевле: {b.Name}");
        else
            Console.WriteLine("  Стоимость одинаковая");

        Console.WriteLine();
    }

    public static void RedirectToPickup(ShippingOption source, Pickup pickup)
    {
        Console.WriteLine($"{source.Name} перенаправляет клиента на самовывоз: {pickup.Address}");
        Console.WriteLine($"Итоговая стоимость: {pickup.CalculateCost()} руб.");
        Console.WriteLine($"Время: {pickup.EstimateDeliveryTime()}");
        Console.WriteLine();
    }
}

DeliveryInteractions.Compare(deliveries[0], deliveries[1]);
DeliveryInteractions.Compare(deliveries[1], deliveries[2]);
DeliveryInteractions.RedirectToPickup(deliveries[1], (Pickup)deliveries[2]);
```
